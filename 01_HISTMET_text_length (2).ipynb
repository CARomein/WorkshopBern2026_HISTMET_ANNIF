## About this notebook

This notebook works entirely with early modern normative texts from Bern (*Policeyordnungen*, the regulatory ordinances or *mandates* by which early modern authorities governed public order, economic life and social conduct). The question it asks, however, comes from elsewhere. The project HISTMET (HIStorical Themes via METadata), funded by the Dutch Research Council (NWO), investigated whether subject metadata can be derived from historical texts *without* labelling them by hand. Its corpus, early modern Dutch provincial resolutions, is divided into short text regions during transcription (a median of some 178 characters, roughly two lines), which were converted into numerical representations with a multilingual language model and grouped automatically. The resulting clusters were coherent, but they sorted by documentary genre rather than by theme: petitions ended up with petitions and resolutions with resolutions, irrespective of what they were about. The project's explanation is that the length of the text regions, rather than the early modern language, governs what the model picks up.

That explanation cannot be tested on the Dutch material, since it has no subject labels. The Bernese mandates can serve as a test case: they are about ten times longer, and each text has been assigned subjects by hand. In this notebook we run the same method twice on the same documents, once on the whole texts and once on a single 178-character piece from each. Documents, method and number of texts stay the same; what changes is the amount of text (and, as the caveats at the end explain, which part of the document it comes from).

This notebook is a simplified teaching version of a test carried out within HISTMET. Steps 5 and 6 explain where it takes shortcuts, so treat its scores as an illustration of the effect rather than as a measurement of it.

### The data

The file `bern_prepared.csv.gz` contains 4,928 of the 4,930 Bernese texts used in Part Two (two texts have no usable transcription). It uses the 2024 transcription (Transkribus Text Titan I, character error rate c. 2.95%), with words split across lines rejoined and layout markup removed. Each text carries one subject from Level 3 of the hierarchical SKOS vocabulary derived from the *Repertorium der Policeyordnungen* (25 categories, e.g. *4.6 Trade and Services*), as selected in C. Annemieke Romein's earlier project *A Game of Thrones* from the labels assigned by hand; many texts carry more than one label in the original data.

# **3. SETTINGS**

This cell fixes the parameters of the experiment and defines a few helper functions; it does not yet compute anything. `CHUNK_CHARS` sets the length of the short pieces (178 characters, the median length of the HISTMET regions), and `SEED` makes the random choices reproducible. The UMAP and HDBSCAN parameters govern the clustering (Step 6).

Agreement is measured with **adjusted mutual information** (AMI): 0 means the clusters are no better than chance at reproducing the hand-assigned subjects, 1 means they reproduce them perfectly. The function `biggest_group_share` guards against a common failure: if one cluster swallows most documents, the AMI is meaningless.

```python
#Step 3. Settings
import re
import warnings
import numpy as np
from sklearn.metrics import adjusted_mutual_info_score

warnings.filterwarnings("ignore", message=".*n_jobs value.*")

CHUNK_CHARS = 178      # median length of the HISTMET regions
MIN_CHARS   = 30
SEED        = 20260911

MIN_CLUSTER_SIZE = 15
MIN_SAMPLES      = 5
SELECTION        = "leaf"
UMAP_DIM         = 50
UMAP_NEIGHBOURS  = 15


def tidy(t):
    return re.sub(r"\s+", " ", str(t)).strip()

def cut_into_pieces(text, target=CHUNK_CHARS):
    """Cut into pieces of roughly `target` characters, never mid-word."""
    words = tidy(text).split(" ")
    pieces, cur, n = [], [], 0
    for w in words:
        add = len(w) if not cur else len(w) + 1
        if cur and n + add > target:
            pieces.append(" ".join(cur)); cur, n = [w], len(w)
        else:
            cur.append(w); n += add
    if cur:
        pieces.append(" ".join(cur))
    return pieces

def agreement(labels, groups):
    """Adjusted mutual information. 0 is chance, 1 is perfect."""
    return float(adjusted_mutual_info_score(labels, groups))

def biggest_group_share(groups):
    """Share of documents in the largest group. If this is high the clustering
    failed and the agreement score means nothing. Check it every time."""
    g = np.asarray(groups)
    real = g[g != -1]
    if real.size == 0:
        return 1.0
    counts = np.bincount(real - real.min())
    return float(counts[counts > 0].max() / g.size)

print("ready")
```

# **4. BUILDING THE TWO VERSIONS**

This cell reads the texts, discards very short ones, and cuts each text into pieces of roughly 178 characters (never mid-word). From every text one piece is chosen at random, so that both versions contain exactly the same number of texts. The output shows the median length of both versions and an example of each.

```python
#Step 4. Build the two versions
docs = pd.read_csv(FNAME)

missing = {"text", "gold"} - set(docs.columns)
if missing:
    raise ValueError(f"{FNAME} has no {sorted(missing)} column. Wrong file.")

if "gold_name" not in docs.columns:
    docs["gold_name"] = docs.gold
    print("Older file, subjects will show as codes rather than names.\n")

docs["text"] = docs.text.map(tidy)
docs = docs.loc[docs.text.str.len() >= MIN_CHARS].reset_index(drop=True)

rng = np.random.default_rng(SEED)
short_text = []
for t in docs.text:
    pieces = [p for p in cut_into_pieces(t) if len(p) >= MIN_CHARS]
    short_text.append(pieces[rng.integers(len(pieces))] if pieces else t[:CHUNK_CHARS])
docs["short"] = short_text

subjects = docs.gold_name.values

print(f"documents        {len(docs)}")
print(f"subjects         {docs.gold_name.nunique()}")
print(f"whole, typical   {int(docs.text.str.len().median())} characters")
print(f"short, typical   {int(docs['short'].str.len().median())} characters")
print()
print("WHOLE :", docs.text.iloc[0][:280], "...")
print()
print("SHORT :", docs["short"].iloc[0])
```

# **5. ENCODING THE TEXT**

A language model converts each text into a vector of 768 numbers (an *embedding*), in such a way that texts with similar content receive similar vectors. We use LaBSE, the multilingual model HISTMET used for the Dutch resolutions. It was built for modern languages, not for early modern German. This notebook is itself a test of how much of the subject it can still pick up. Loading the model and encoding both versions takes a few minutes on a GPU.

It is important to note that the model reads at most 256 *tokens* (words or word fragments) per text, which for these texts amounts to roughly 896 characters; anything beyond is ignored. Since some 90 per cent of the mandates are longer than that, the "whole documents" in this notebook are, strictly speaking, the opening portion of each document. One way round this is to turn every 178-character piece of a mandate into a vector and average them, so that the whole text is covered. This notebook keeps the simpler and faster version.

```python
#Step 5. Encode the text
from sentence_transformers import SentenceTransformer
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)
if device == "cpu":
    print("No GPU, so this takes roughly 15 minutes.")
    print("In Colab: Runtime, Change runtime type, T4 GPU.")

model = SentenceTransformer("sentence-transformers/LaBSE", device=device)
model.max_seq_length = 256

print("\nwhole documents")
vec_whole = model.encode(docs.text.tolist(), batch_size=64, show_progress_bar=True,
                         convert_to_numpy=True, normalize_embeddings=True)
print("short pieces")
vec_short = model.encode(docs["short"].tolist(), batch_size=64, show_progress_bar=True,
                         convert_to_numpy=True, normalize_embeddings=True)
print(f"\n{vec_whole.shape[1]} numbers per text")
```

# **6. CLUSTERING AND SCORING**

Clustering happens in two stages. UMAP first reduces the 768 numbers per text to 50, which makes the structure of the data easier to detect; HDBSCAN then groups texts that lie densely together. HDBSCAN decides the number of clusters itself and may label texts as *noise* (`-1`) when they fit no cluster; those texts are left out of the agreement score. For each version, the output reports the number of clusters, the share of texts in the largest cluster, the share left unassigned (and therefore not scored), and the AMI with the hand-assigned subjects.

The setting `SELECTION = "leaf"` makes HDBSCAN prefer many small, fine-grained clusters. With HDBSCAN's default (*excess of mass*), the clustering of short or truncated texts collapses into a single cluster holding almost all points, which makes the agreement score meaningless.

Before you run the next cell, make a guess. Which version will agree better with the hand-assigned subjects, and by how much?

```python
#Step 6. Cluster and score
import umap, hdbscan

def find_groups(vectors, n_dims=UMAP_DIM):
    reduced = umap.UMAP(n_components=n_dims, n_neighbors=UMAP_NEIGHBOURS,
                        metric="cosine", random_state=SEED).fit_transform(vectors)
    return hdbscan.HDBSCAN(min_cluster_size=MIN_CLUSTER_SIZE, min_samples=MIN_SAMPLES,
                           cluster_selection_method=SELECTION).fit_predict(reduced)

groups_whole = find_groups(vec_whole)
groups_short = find_groups(vec_short)

results = {}
for name, groups in (("Whole documents", groups_whole), ("Short pieces", groups_short)):
    keep = groups != -1
    score = agreement(subjects[keep], groups[keep])
    share = biggest_group_share(groups)
    unassigned = float((~keep).mean())
    results[name] = {"score": score, "biggest": share, "unassigned": unassigned,
                     "n_groups": int(len(set(groups[keep])))}
    warn = "   <- clustering failed, ignore the score" if share > 0.5 else ""
    print(f"{name:<18} groups {results[name]['n_groups']:>4}   biggest {share:>5.0%}"
          f"   unassigned {unassigned:>5.0%}   agreement {score:.4f}{warn}")
```

# **WHAT IT SHOWS**

Cut down to the length of the HISTMET regions, the Bernese mandates barely cluster by subject at all. Left whole, they score several times higher. Since the documents, the model and the settings stayed the same, the amount of text appears to be doing the work. The map tells the same story more directly: on the left the subjects occupy recognisable areas (*Matters of Religion* most clearly), on the right they are largely mixed together.

This is what HISTMET found for the Dutch resolutions, whose text regions are just as short and whose clusters sorted by genre rather than by theme. Here a different territory, language and kind of text behave in the same way once the texts are equally short. This suggests that genre wins over theme because of the length of the unit of text, not because of the method or of early modern Dutch.

Four caveats deserve mention. First, the whole-text version always contains the opening of the document, where the title and the clearest statement of the subject tend to appear. The short piece comes from a random position. So length and position are partly mixed up (Step 10 looks at this). Second, a 178-character piece keeps the subject of the mandate it was cut from, but may not mention that subject itself. This makes the drop in agreement look larger. Third, an AMI of about 0.25 is still modest. Even with longer texts, the clusters capture subject only in part. Fourth, HDBSCAN leaves many texts unassigned (in our run about 60 to 70 per cent of each version), and the score only covers the texts that were assigned. Step 6 prints this share for your own run.

# **9. INSIDE THE CLUSTERS (OPTIONAL)**

A score is abstract, so this step opens up the three biggest clusters of each version. For each cluster it shows how many texts it holds, which hand-assigned subject is most common in it and what share of the cluster that subject covers, and the start of one text from it.

The last line per version averages that share over all clusters, weighted by their size, and sets it beside a random baseline. The baseline shuffles the subjects over the same clusters, so it shows what the share would be if the clusters had nothing to do with subject. Small clusters look purer by chance alone, which is why the baseline matters.

If length is what matters, the whole-document clusters should sit further above their random baseline than the short-piece clusters. Run the cell and see whether that holds. It takes a few seconds, and you can skip it if time is short.

```python
#Step 9 (optional). Look inside the biggest clusters
def dominant_share(groups, labels=None):
    """Per cluster, the share of its texts carrying its most common subject,
    averaged over clusters and weighted by cluster size. Unassigned texts are left out."""
    g = pd.Series(groups)
    s = pd.Series(subjects if labels is None else labels)
    keep = g != -1
    if not keep.any():
        return float("nan")
    top = s[keep].groupby(g[keep]).agg(lambda x: x.value_counts().iloc[0])
    return float(top.sum() / keep.sum())

def look_inside(groups, texts, n_clusters=3, width=110):
    g, s = pd.Series(groups), pd.Series(subjects)
    for cid, size in g[g != -1].value_counts().head(n_clusters).items():
        mix = s[g == cid].value_counts()
        example = texts[g[g == cid].index[0]]
        print(f"  cluster {cid:>3}   {size:>4} texts   most common subject {mix.index[0]} ({mix.iloc[0] / size:.0%})")
        print(f"                e.g. {example[:width]}...")
    shuffled = np.random.default_rng(SEED).permutation(subjects)
    print(f"  on average the most common subject covers {dominant_share(groups):.0%} of a cluster"
          f"   (random baseline {dominant_share(groups, shuffled):.0%})")

for name, groups, texts in (("Whole documents", groups_whole, docs.text.tolist()),
                            ("Short pieces", groups_short, docs["short"].tolist())):
    print(name)
    look_inside(groups, texts)
    print()
```

# **10. HOW MUCH TEXT IS ENOUGH? (OPTIONAL)**

So far we compared two lengths. This step adds pieces of about 100, 400 and 800 characters, one per mandate and chosen the same way as in Step 4. It then plots the agreement for all five versions. It takes a few minutes on a GPU.

Look for two things. First, where does the line start to rise? Second, how do the 800-character pieces compare with the whole documents? The model reads only about the first 900 characters of a whole document (Step 5), so both give it about the same amount of text. If the whole documents still score clearly higher, then the opening of a mandate says more about its subject than a random stretch from the middle.

A hollow point means one cluster took more than half of the texts. That score means nothing.

Our result, for reference:

![Expected curve](https://raw.githubusercontent.com/CARomein/WorkshopBern2026_HISTMET_ANNIF/master/histmet/bern_length_curve.png)


```python
#Step 10 (optional). Agreement at different piece lengths
EXTRA_LENGTHS = [100, 400, 800]   # characters; 178 and the whole documents are reused from above

def one_piece_per_text(target, seed=SEED):
    """One random piece of roughly `target` characters from every text, as in Step 4."""
    r = np.random.default_rng(seed)
    out = []
    for t in docs.text:
        pieces = [p for p in cut_into_pieces(t, target) if len(p) >= MIN_CHARS]
        out.append(pieces[r.integers(len(pieces))] if pieces else t[:target])
    return out

def score_version(groups):
    keep = groups != -1
    return {"score": agreement(subjects[keep], groups[keep]) if keep.any() else float("nan"),
            "biggest": biggest_group_share(groups), "unassigned": float((~keep).mean())}

curve = {f"{CHUNK_CHARS}": results["Short pieces"], "whole": results["Whole documents"]}
for n in EXTRA_LENGTHS:
    texts = one_piece_per_text(n)
    vec = model.encode(texts, batch_size=64, show_progress_bar=False,
                       convert_to_numpy=True, normalize_embeddings=True)
    curve[f"{n}"] = score_version(find_groups(vec))
    print(f"{n:>4} characters   done")

order = [str(n) for n in sorted(EXTRA_LENGTHS + [CHUNK_CHARS])] + ["whole"]
print()
for k in order:
    r = curve[k]
    flag = "   <- clustering failed, ignore the score" if r["biggest"] > 0.5 else ""
    print(f"{k:>6}   agreement {r['score']:.4f}   biggest {r['biggest']:>4.0%}   unassigned {r['unassigned']:>4.0%}{flag}")

fig, ax = plt.subplots(figsize=(7, 4.2), facecolor=SURFACE)
ax.set_facecolor(SURFACE)
xs = list(range(len(order)))
ys = [curve[k]["score"] for k in order]
ax.plot(xs[:-1], ys[:-1], color=ORANGE, lw=2, zorder=2)
for x, k, y in zip(xs, order, ys):
    colour = BLUE if k == "whole" else ORANGE
    failed = curve[k]["biggest"] > 0.5
    ax.scatter([x], [y], s=70, zorder=3, color=SURFACE if failed else colour, edgecolor=colour, lw=2)
    ax.text(x, y + max(ys) * 0.04, f"{y:.3f}", ha="center", va="bottom", color=INK, fontsize=10)
ax.set_xticks(xs)
ax.set_xticklabels([f"{k} chars" if k != "whole" else "whole\ndocuments" for k in order])
ax.set_ylabel("agreement with hand-assigned subject", color=INK_SOFT, fontsize=10)
ax.set_title("How much text does the method need?", color=INK, fontsize=13, pad=14, loc="left")
ax.set_ylim(0, max(ys) * 1.3)
ax.grid(axis="y", color="#e6e5e1", lw=0.8, zorder=0)
ax.set_axisbelow(True)
for side in ("top", "right", "left"):
    ax.spines[side].set_visible(False)
ax.spines["bottom"].set_color("#d8d7d2")
ax.tick_params(colors=INK_SOFT, length=0, labelsize=9)
plt.tight_layout()
fig.savefig("bern_length_curve.png", dpi=200, facecolor=SURFACE, bbox_inches="tight")
plt.show()
print("saved bern_length_curve.png")
```

In our run the score rose with every step in length, and no clustering failed. The 800-character pieces reached only about half the score of the whole documents, although the model reads roughly the same amount of text in both. So the part of the text matters as well as the amount. The opening of a mandate, where the title usually sits, seems to say the most about its subject.
